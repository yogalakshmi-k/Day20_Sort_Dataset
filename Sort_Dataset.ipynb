```python

import pandas as pd

# Create Student Performance Dataset
data = {
    'Student': ['Arun', 'Bala', 'Divya', 'Kavi', 'Meena', 'Ravi'],
    'Age': [21, 20, 22, 21, 20, 23],
    'Marks': [75, 45, 90, 60, 85, 45],
    'Department': ['MCA', 'MBA', 'MCA', 'BCA', 'MBA', 'BCA']
}

# Convert data into a DataFrame
df = pd.DataFrame(data)

# Display original dataset
print("ORIGINAL DATASET")
print(df.to_string(index=False))

# 1. Ascending Order
print("\n1. ASCENDING ORDER (Marks: Low to High)")
ascending = df.sort_values(by='Marks', ascending=True)
print(ascending.to_string(index=False))

# 2. Descending Order
print("\n2. DESCENDING ORDER (Marks: High to Low)")
descending = df.sort_values(by='Marks', ascending=False)
print(descending.to_string(index=False))

# 3. Multiple Column Sorting
print("\n3. MULTIPLE COLUMN SORTING")
multiple_sort = df.sort_values(
    by=['Department', 'Marks'],
    ascending=[True, False]
)
print(multiple_sort.to_string(index=False))

# 4. Sort by Age
print("\n4. SORT BY AGE (Youngest to Oldest)")
age_sort = df.sort_values(by='Age', ascending=True)
print(age_sort.to_string(index=False))

# 5. Top 3 Students
print("\n5. TOP 3 STUDENTS")
top_students = df.sort_values(
    by='Marks', ascending=False
).head(3)
print(top_students.to_string(index=False))

print("\nTASK COMPLETED SUCCESSFULLY!")
```
